 **2M1.** Recall the globe tossing model from the chapter. Compute and plot the grid approximate posterior distribution for each of the following sets of observations. In each case, assume a uniform prior for $p$

  1. W, W, W
  2. W, W, W, L
  3. L, W, W, L, W, W, W



```python
# 1 W, W, W,
import torch
import pyro
import pyro.distributions as dist
import seaborn as sns

p_grid = torch.linspace(start=0, end=1, steps=50)
prior = torch.tensor(1.).repeat(50)
likelihood = dist.Binomial(total_count=3, probs=p_grid).log_prob(torch.tensor(3.)).exp()
unstd_posterior = likelihood * prior
posterior = unstd_posterior / unstd_posterior.sum()

ax = sns.lineplot(x=p_grid, y=posterior, marker="o")
ax.set(xlabel="probability of water", ylabel="posterior probability");
```

```python
# 2 W, W, W, L

likelihood = dist.Binomial(total_count=4, probs=p_grid).log_prob(torch.tensor(3.)).exp()
unstd_posterior = likelihood * prior
posterior = unstd_posterior / unstd_posterior.sum()

ax = sns.lineplot(x=p_grid, y=posterior, marker="o")
ax.set(xlabel="probability of water", ylabel="posterior probability");
```
